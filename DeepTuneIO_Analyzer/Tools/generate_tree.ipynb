````python
import os
import json

try:
    import yaml
except ImportError:
    yaml = None  # YAML es opcional

def build_tree_dict(path):
    """
    Construye una estructura jerárquica en forma de diccionario.

    Parámetros
    ----------
    path : str
        Ruta del directorio raíz.

    Retorna
    -------
    dict
        Estructura del árbol en formato dict.
    """
    tree = {"name": os.path.basename(path), "type": "directory", "children": []}

    try:
        items = sorted(os.listdir(path))
    except Exception as e:
        tree["children"].append({"name": f"[ERROR: {e}]", "type": "error"})
        return tree

    for item in items:
        full_path = os.path.join(path, item)
        if os.path.isdir(full_path):
            tree["children"].append(build_tree_dict(full_path))
        else:
            tree["children"].append({"name": item, "type": "file"})

    return tree


def build_tree_text(path, prefix=""):
    """
    Genera un árbol estilo Unix en formato texto.

    Parámetros
    ----------
    path : str
        Ruta del directorio raíz.
    prefix : str
        Prefijo visual para alinear el árbol.

    Retorna
    -------
    list[str]
        Líneas del árbol.
    """
    lines = []
    try:
        items = sorted(os.listdir(path))
    except Exception as e:
        return [prefix + f"└── [ERROR: {e}]"]

    total = len(items)

    for i, item in enumerate(items):
        full_path = os.path.join(path, item)
        connector = "└── " if i == total - 1 else "├── "
        lines.append(prefix + connector + item)

        if os.path.isdir(full_path):
            extension = "    " if i == total - 1 else "│   "
            lines.extend(build_tree_text(full_path, prefix + extension))

    return lines


def export_txt(tree_lines, output_path):
    """Guarda el árbol en formato TXT."""
    with open(output_path, "w", encoding="utf-8") as f:
        f.write("\n".join(tree_lines))


def export_json(tree_dict, output_path):
    """Guarda el árbol en formato JSON."""
    with open(output_path, "w", encoding="utf-8") as f:
        json.dump(tree_dict, f, indent=4, ensure_ascii=False)


def export_yaml(tree_dict, output_path):
    """Guarda el árbol en formato YAML (si PyYAML está disponible)."""
    if yaml is None:
        print("PyYAML no está instalado. No se generará YAML.")
        return
    with open(output_path, "w", encoding="utf-8") as f:
        yaml.dump(tree_dict, f, allow_unicode=True)


def export_markdown(tree_lines, output_path):
    """Guarda el árbol en formato Markdown."""
    with open(output_path, "w", encoding="utf-8") as f:
        f.write("# Árbol de directorios\n\n```\n")
        f.write("\n".join(tree_lines))
        f.write("\n```\n")


def generate_all_exports(root_path, output_dir=None):
    """
    Genera todos los formatos de exportación.

    Parámetros
    ----------
    root_path : str
        Ruta del directorio a analizar.
    output_dir : str | None
        Carpeta donde guardar los archivos. Si es None, usa root_path.
    """
    if output_dir is None:
        output_dir = root_path

    print("Generando árbol...")

    # TXT
    tree_text = build_tree_text(root_path)
    export_txt(tree_text, os.path.join(output_dir, "IOForger_Bench_tree.txt"))

    # JSON
    tree_dict = build_tree_dict(root_path)
    export_json(tree_dict, os.path.join(output_dir, "IOForger_Bench_tree.json"))

    # YAML
    export_yaml(tree_dict, os.path.join(output_dir, "IOForger_Bench_tree.yaml"))

    # Markdown
    export_markdown(tree_text, os.path.join(output_dir, "IOForger_Bench_tree.md"))

    print("Exportaciones completadas.")


if __name__ == "__main__":
    root = r"F:\NoteBooks\PHD_THESIS\Tools\IOForger_Bench"

    if not os.path.exists(root):
        print("La ruta no existe:", root)
    else:
        generate_all_exports(root)
````

````python
import os

def build_tree(root_path, prefix=""):
    """
    Recorre recursivamente un directorio y construye una representación
    en forma de árbol usando caracteres estilo Unix.

    Parámetros
    ----------
    root_path : str
        Ruta del directorio raíz a analizar.
    prefix : str
        Prefijo usado para alinear visualmente los elementos del árbol.

    Retorna
    -------
    list[str]
        Lista de líneas que representan el árbol completo.
    """
    tree_lines = []

    try:
        items = sorted(os.listdir(root_path))
    except PermissionError:
        return [prefix + "└── [ACCESO DENEGADO]"]
    except FileNotFoundError:
        return [prefix + "└── [NO ENCONTRADO]"]

    total = len(items)

    for index, item in enumerate(items):
        full_path = os.path.join(root_path, item)
        connector = "└── " if index == total - 1 else "├── "
        tree_lines.append(prefix + connector + item)

        # Si es un directorio, recursión
        if os.path.isdir(full_path):
            extension = "    " if index == total - 1 else "│   "
            tree_lines.extend(build_tree(full_path, prefix + extension))

    return tree_lines


def print_tree(tree_lines):
    """
    Imprime el árbol generado línea por línea.
    """
    for line in tree_lines:
        print(line)


def save_tree(tree_lines, output_path):
    """
    Guarda el árbol en un archivo de texto.

    Parámetros
    ----------
    tree_lines : list[str]
        Líneas del árbol generado.
    output_path : str
        Ruta del archivo donde se guardará el resultado.
    """
    try:
        with open(output_path, "w", encoding="utf-8") as f:
            f.write("\n".join(tree_lines))
        print(f"Árbol guardado en: {output_path}")
    except Exception as e:
        print(f"Error al guardar el archivo: {e}")


if __name__ == "__main__":
    # Ruta del proyecto
    root = r"F:\NoteBooks\PHD_THESIS\Tools\IOForger_Bench"

    if not os.path.exists(root):
        print("La ruta no existe:", root)
    else:
        print("Generando árbol del directorio...\n")
        tree = build_tree(root)

        # Mostrar en pantalla
        print_tree(tree)

        # Guardar en archivo (opcional)
        output_file = os.path.join(root, "IOForger_Bench_tree.txt")
        save_tree(tree, output_file)
````
