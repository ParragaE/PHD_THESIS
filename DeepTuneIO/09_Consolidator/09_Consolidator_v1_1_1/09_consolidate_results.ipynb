# DeepTuneIO — Module 09: Consolidator v1.1.1

## Purpose
Integrate one row per JobID while preserving source provenance and separating analytical readiness from optional resource/training metadata.

## Inputs
Required: Indexer, Parser, DXT and PERF. Optional: SEFF (SLURM/resource metadata) and Training summary.

## Processing
Indexer anchor → LEFT JOIN sources → canonical fields → consistency checks → source availability → analytical readiness → outputs.

## Outputs
`results/Consolidated/Execution/`, `Validation/`, and `Metadata/`. See `Documentation/MODULE_09_TECHNICAL_DOCUMENTATION.md` and `INPUT_OUTPUT_MATRIX.md`.


```python
from pathlib import Path
import sys
from importlib.util import spec_from_file_location, module_from_spec

MODULE_DIR = Path("./Module").resolve()
MODULE09_FILE = MODULE_DIR / "09_consolidate_results.py"

spec = spec_from_file_location("deeptuneio_module09", MODULE09_FILE)
module09 = module_from_spec(spec)
sys.modules[spec.name] = module09
spec.loader.exec_module(module09)

print("Module 09:", MODULE09_FILE)
print("Version  :", module09.VERSION)
```

## Select campaign by index\n

```python
CAMPAIGNS = {
    "DeepGalaxy": {
        "DG_bw512_f3c5x64": {
            "scenarios": ["lustre_1ost", "lustre_2ost", "lustre_4ost", "nfs_SnGPU_e1"],
            "file_format": ".hdf5",
        },
    },
    "DLIOv1": {
        "HDF5_64bs": {"scenarios": ["lustre_ost"], "file_format": ".h5"},
        "NPZ_64bs": {"scenarios": ["lustre_ost"], "file_format": ".npz"},
        "TFRecord_256kts_64bs": {"scenarios": ["lustre_ost"], "file_format": ".tfrecords"},
        "TFRecord_1mts_64bs": {"scenarios": ["lustre_ost"], "file_format": ".tfrecords"},
    },
}
APPLICATION_POS = 1
DATASET_POS = 1
SCENARIO_POS = 3

APPLICATION = list(CAMPAIGNS)[APPLICATION_POS-1]
DATASET_GROUP = list(CAMPAIGNS[APPLICATION])[DATASET_POS-1]
SCENARIO = CAMPAIGNS[APPLICATION][DATASET_GROUP]["scenarios"][SCENARIO_POS-1]
FILE_FORMAT = CAMPAIGNS[APPLICATION][DATASET_GROUP]["file_format"]

print("Application   :", APPLICATION)
print("Dataset group :", DATASET_GROUP)
print("Scenario      :", SCENARIO)
print("File format   :", FILE_FORMAT)
```

## Build campaign root\n

```python
PROJECT_ROOT = Path(r"F:\\NoteBooks\\CORA_DR_UAB").resolve()
ARTICLE_ID = "Article_01_Parallel_IO_Analysis"
ARTICLE_ROOT = PROJECT_ROOT / ARTICLE_ID
CAMPAIGN_ROOT = ARTICLE_ROOT / "Data" / APPLICATION / DATASET_GROUP / SCENARIO

print("Project root  :", PROJECT_ROOT)
print("Article root  :", ARTICLE_ROOT)
print("Campaign root :", CAMPAIGN_ROOT)
print("Exists        :", CAMPAIGN_ROOT.exists())
```

## Resolve Module 09 paths and discover inputs\n

```python
args = module09.parse_args([
    "--campaign-root", str(CAMPAIGN_ROOT),
    "--application", APPLICATION,
    "--dataset-group", DATASET_GROUP,
    "--scenario", SCENARIO,
    "--file-format", FILE_FORMAT,
    "--event", "JSC24",
    "--module-root", str(MODULE_DIR),
])
paths = module09.resolve_campaign_paths(args)
for key, value in paths.items():
    print(f"{key:22s}: {value}")

print("\nInputs:")
inputs = module09.discover_inputs(args, paths)
for key, value in inputs.items():
    print(f"{key:10s}: {value}")
```

## Execute Module 09\n

```python
return_code = module09.main([
    "--campaign-root", str(CAMPAIGN_ROOT),
    "--application", APPLICATION,
    "--dataset-group", DATASET_GROUP,
    "--scenario", SCENARIO,
    "--file-format", FILE_FORMAT,
    "--event", "JSC24",
    "--module-root", str(MODULE_DIR),
])
print("Return code:", return_code)
```

## Inspect generated results\n

```python
CONSOLIDATED_DIR = CAMPAIGN_ROOT / "results" / "Consolidated"
for p in sorted(CONSOLIDATED_DIR.rglob("*")) if CONSOLIDATED_DIR.exists() else []:
    if p.is_file():
        print(" -", p.relative_to(CONSOLIDATED_DIR))
```

## Inspect readiness and metadata status

This check distinguishes core analytical readiness from optional Training/SEFF availability.


```python
import pandas as pd
validation_files = sorted((CONSOLIDATED_DIR / 'Validation').glob('consolidation_validation_*.csv'))
if validation_files:
    validation_df = pd.read_csv(validation_files[-1])
    cols = [c for c in ['job_id','consolidation_status','source_completeness_status','performance_analysis_ready','resource_metadata_status','training_metadata_status','configuration_consistency'] if c in validation_df.columns]
    display(validation_df[cols])
else:
    print('No validation output found.')
```
