```python
import os
import re
import pandas as pd

# Función para convertir bytes a una unidad legible (e.g., MiB, GiB)
def formatear_bytes(bytes_val):
    if bytes_val is None: return None
    for unidad in ['B', 'KiB', 'MiB', 'GiB', 'TiB']:
        if bytes_val < 1024:
            return f"{bytes_val:.2f} {unidad}"
        bytes_val /= 1024
    return f"{bytes_val:.2f} PiB"

def consolidar_logs_dlio_pro(lista_rutas):
    # Regex para extraer datos clave
    patron_exe = re.compile(r"# exe: (.*)")
    patron_perf = re.compile(r"agg_perf_by_slowest:\s+([\d\.]+)")
    patron_time = re.compile(r"# run time: ([\d\.]+)")
    patron_bytes = re.compile(r"# total_bytes: ([\d\.]+)")
    patron_io = re.compile(r"shared files: time_by_slowest: ([\d\.]+)")
    patron_args = re.compile(r"(-\w+)\s+([^\s]+)")
    
    lista_resultados = []
    
    for ruta in lista_rutas:
        if not os.path.exists(ruta): continue
        for root, _, files in os.walk(ruta):
            for file in [f for f in files if f.endswith("_perf.txt")]:
                with open(os.path.join(root, file), 'r', encoding='utf-8', errors='ignore') as f:
                    contenido = f.read()
                    
                    match_exe = patron_exe.search(contenido)
                    if match_exe:
                        datos = {'archivo': file, 'configuracion': os.path.basename(ruta)}
                        
                        # Extraer argumentos
                        datos.update(dict(patron_args.findall(match_exe.group(1))))
                        
                        # Extraer métricas adicionales
                        m_time = patron_time.search(contenido)
                        m_bytes = patron_bytes.search(contenido)
                        m_io = patron_io.search(contenido)
                        m_perf = patron_perf.search(contenido)
                        
                        datos['run_time_s'] = float(m_time.group(1)) if m_time else None
                        datos['io_time_s'] = float(m_io.group(1)) if m_io else None
                        datos['total_bytes'] = int(m_bytes.group(1)) if m_bytes else None
                        datos['total_bytes_fmt'] = formatear_bytes(datos['total_bytes'])
                        datos['perf_mib_s'] = float(m_perf.group(1)) if m_perf else None
                        
                        lista_resultados.append(datos)
    
    return pd.DataFrame(lista_resultados)

# --- EJECUCIÓN ---
rutas = [
    r"F:\NoteBooks\PHD_THESIS\Papers\The_Journal_of_Supercomputing_2025\Data\DLIOv1\HDF5_64bs",
    r"F:\NoteBooks\PHD_THESIS\Papers\The_Journal_of_Supercomputing_2025\Data\DLIOv1\NPZ_64bs",
    r"F:\NoteBooks\PHD_THESIS\Papers\The_Journal_of_Supercomputing_2025\Data\DLIOv1\TFRecord_1mts_64bs",
    r"F:\NoteBooks\PHD_THESIS\Papers\The_Journal_of_Supercomputing_2025\Data\DLIOv1\TFRecord_256kts_64bs"
]

df_deepio = consolidar_logs_dlio_pro(rutas)

# Limpieza de nombres de columnas
df_deepio.columns = [c.replace('-', '') for c in df_deepio.columns]

# Mostrar tabla final
display(df_deepio[['archivo', 'configuracion', 'run_time_s', 'io_time_s', 'total_bytes_fmt', 'perfmibs']])
```

```python
import os
import re
import pandas as pd

def consolidar_logs_dlio(lista_rutas):
    patron_exe = re.compile(r"# exe: (.*)")
    patron_perf = re.compile(r"agg_perf_by_slowest:\s+([\d\.]+)")
    patron_args = re.compile(r"(-\w+)\s+([^\s]+)")
    
    lista_resultados = []
    
    for ruta in lista_rutas:
        if not os.path.exists(ruta):
            print(f"Advertencia: La ruta no existe -> {ruta}")
            continue
            
        for root, _, files in os.walk(ruta):
            for file in [f for f in files if f.endswith("_perf.txt")]:
                with open(os.path.join(root, file), 'r', encoding='utf-8', errors='ignore') as f:
                    contenido = f.read()
                    
                    match_exe = patron_exe.search(contenido)
                    if match_exe:
                        datos = {'archivo': file, 'ruta_padre': os.path.basename(ruta)}
                        
                        # Extraer argumentos dinámicamente
                        argumentos = dict(patron_args.findall(match_exe.group(1)))
                        datos.update(argumentos)
                        
                        # Extraer rendimiento
                        match_perf = patron_perf.search(contenido)
                        if match_perf:
                            datos['perf_mib_s'] = float(match_perf.group(1))
                        
                        lista_resultados.append(datos)
    
    return pd.DataFrame(lista_resultados)

# --- CONFIGURACIÓN DE RUTAS ---
rutas_a_procesar = [
    r"F:\NoteBooks\PHD_THESIS\Papers\The_Journal_of_Supercomputing_2025\Data\DLIOv1\HDF5_64bs",
    r"F:\NoteBooks\PHD_THESIS\Papers\The_Journal_of_Supercomputing_2025\Data\DLIOv1\NPZ_64bs",
    r"F:\NoteBooks\PHD_THESIS\Papers\The_Journal_of_Supercomputing_2025\Data\DLIOv1\TFRecord_1mts_64bs",
    r"F:\NoteBooks\PHD_THESIS\Papers\The_Journal_of_Supercomputing_2025\Data\DLIOv1\TFRecord_256kts_64bs"
]

# --- EJECUCIÓN ---
df_deepio = consolidar_logs_dlio(rutas_a_procesar)

# Limpieza opcional: quitar guiones de los nombres de columna para facilitar el uso
df_deepio.columns = [c.replace('-', '') for c in df_deepio.columns]

display(df_deepio)
```

```python
import os
import re
import pandas as pd

def extraer_datos_dlio_dinamico(directorio_base):
    # Patrón para el comando y para el rendimiento
    patron_exe = re.compile(r"# exe: (.*)")
    patron_perf = re.compile(r"agg_perf_by_slowest:\s+([\d\.]+)")
    
    # Este patrón captura cualquier flag (-x) y su valor asociado
    # Funciona para -f hdf5, -bs 64, -df /ruta/larga, etc.
    patron_args = re.compile(r"(-\w+)\s+([^\s]+)")
    
    resultados = []
    
    for root, _, files in os.walk(directorio_base):
        for file in [f for f in files if f.endswith("_perf.txt")]:
            ruta_completa = os.path.join(root, file)
            
            with open(ruta_completa, 'r', encoding='utf-8', errors='ignore') as f:
                contenido = f.read()
                
                # Extraer comando
                match_exe = patron_exe.search(contenido)
                if match_exe:
                    comando = match_exe.group(1).strip()
                    
                    # Diccionario base con nombre de archivo
                    datos = {'archivo': file}
                    
                    # 1. Extraer todos los parámetros dinámicamente
                    argumentos = dict(patron_args.findall(comando))
                    datos.update(argumentos)
                    
                    # 2. Extraer rendimiento
                    match_perf = patron_perf.search(contenido)
                    if match_perf:
                        datos['perf_mib_s'] = float(match_perf.group(1))
                    
                    resultados.append(datos)
    
    return pd.DataFrame(resultados)

# --- Ejecución ---
df_deepio = extraer_datos_dlio_dinamico('.')

# Mostrar resultado
pd.set_option('display.max_columns', None)
display(df_deepio.head())
```
